## Model Invocation

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_BASE_URL"]=os.getenv("BEDROCK_MANTLE_BASE_URL")
os.environ["OPENAI_API_KEY"]=os.getenv("BEDROCK_MANTLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("openai.gpt-oss-20b", model_provider="openai")
model
```

```python
response = model.invoke("Hello, how are you?")
response.content
```

# Streaming & Batch

## Streaming

```python
stream_response = model.stream("Write me a para on AI Agents")
stream_response
```

```python
for chunk in stream_response:
    print(chunk.text, "|", flush=True)
```

## Batch

```python
model.batch(
    [
        "Why is RED?",
        "Why do birds fly?"
    ],
    config={
        "max_concurrency": 5
    }
)
```
